# 1. create an object and access

```python
from pydantic import BaseModel

class User(BaseModel):
  uid : int
  name : str

sample = User(uid = 10 , name = "Suraj")
print(sample)
print(sample.uid, type(sample.uid))
print(sample.name, type(sample.name))
```

# 7. Custom Data Validation

```python
from pydantic import BaseModel, Field, field_validator
class User(BaseModel):
  age : int = Field(gt=17, lt=40)
  name : str = Field(min_length=3, max_length=20)
  # email : Optional[str] = None
  email : str = None

  @field_validator("email")
  @classmethod
  def email_data_validator(cls,value):
    if not value.endswith("@microdegree.com"):
      raise ValueError("Sorry Your not eligible for creating this object")
    return value
```

```python
# with wrong input
sample = User(age =20, name="suraj", email="sample@gmail.com")
print(sample)
```

```python
# with correct input
sample = User(age =20, name="suraj", email="sample@microdegree.com")
print(sample)
```

# 7. Supports Description for each field

```python
from pydantic import BaseModel, Field

class User(BaseModel):
  age : int = Field(gt=17, lt=40, description="Agen of an employee")
  name : str = Field(min_length=3, max_length=20 ,description= "Full name of an employee")
  # email : Optional[str] = None
  email : str = None
```

# Nested Pydantic Model

```python
from pydantic import BaseModel
## You can use pydantic model as another data type
class Address(BaseModel):
  city : str
  country : str

class User(BaseModel):
  uid : int
  name : str
  address : Address
```

```python
addr = Address(city="Bengaluru", country="India")
print(addr)
```

```python
usr1 = User(uid=99, name="Vinay", address = addr)
usr1
```

```python
usr2 = User(uid=99, name="Vinay", address={"city":"Pune", "country":"India"})
usr2
```

```python
usr2.address.city
```

```python
usr1.address.country
```

# Default Factory

```python
from uuid import uuid4
from pydantic import BaseModel, Field

class User(BaseModel):
  uid : str = Field(default_factory= lambda : str(uuid4()))
  name : str

usr1 = User(name="Vijay")
usr1
```

```python
from uuid import uuid4
from pydantic import BaseModel, Field

def get_uid():
  return "ThisIsFromFunction"

class User(BaseModel):
  uid : str = Field(default_factory= get_uid)
  name : str

usr1 = User(name="Vijay")
usr1
```

# Pydantic Example


```python
from pydantic import BaseModel, Field , field_validator

def sample():
  uid = uuid4()
  return str(uid)

class Adress(BaseModel):
  city : str
  country : str

class User(BaseModel):
  uid : str = Field(default_factory=sample)       # Default Factory
  name : str = Field(min_length=3, max_length=26)   # Type and Basic Data Validation for str
  age : int = Field(gt=17, lt=45)   # Type and Basic Data Validation for int
  email : str = None                # Make paramter as optional and check custom validation if value is provided
  address :Adress                   # Using Nested Object Concept

  @field_validator("email")
  @classmethod
  def check_email(cls, evalue):
    if not evalue.endswith("@gmail.com"):
      raise ValueError("Not a Valid Gmail ")
    return evalue
```
