```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('google/embeddinggemma-2')

user_query = "What is Azure Functions?"
query_embedding = model.encode(user_query)
chunk_embeddings = model.encode(chunks)

print("Query Embedding:", query_embedding)
print("Chunk Embeddings:", chunk_embeddings)
print("Query Embedding Shape:", query_embedding.shape)
print("Chunk Embeddings Shape:", chunk_embeddings.shape)
```

```python
import numpy as np
import pandas as pd
from sentence_transformers import util

cosine_similarities = util.cos_sim(query_embedding, chunk_embeddings)
print("Cosine Similarities:", cosine_similarities)

scores = cosine_similarities[0].tolist()

cosine_ranks = sorted(
    range(len(scores)),
    key=lambda i: scores[i],
    reverse=True
)

for rank, index in enumerate(cosine_ranks, start=1):
    print(f"Rank: {rank}")
    print(f"Score: {scores[index]:.4f}")
    print(f"Chunk: {chunks[index]}")
    print("-" * 50)

dot_scores = util.dot_score(query_embedding, chunk_embeddings)
print("Dot Scores:", dot_scores)

scores = dot_scores[0].tolist()

dot_ranks = sorted(
    range(len(scores)),
    key=lambda i: scores[i],
    reverse=True
)

for rank, index in enumerate(dot_ranks, start=1):
    print(f"Rank: {rank}")
    print(f"Score: {scores[index]:.4f}")
    print(f"Chunk: {chunks[index]}")
    print("-" * 50)

#print chunks from cosine similarity and dot product similarity and compare the results rank in table format
# 3. Map each chunk index to its rank
cosine_rank_map = {idx: rank for rank, idx in enumerate(cosine_ranks, start=1)}
dot_rank_map = {idx: rank for rank, idx in enumerate(dot_ranks, start=1)}

# 4. Create comparison table
comparison = pd.DataFrame({
    "Chunk": chunks,
    "Cosine Score": cosine_similarities[0].tolist(),
    "Cosine Rank": [cosine_rank_map[i] for i in range(len(chunks))],
    "Dot Product Score": dot_scores[0].tolist(),
    "Dot Product Rank": [dot_rank_map[i] for i in range(len(chunks))]
})

# 5. Sort by cosine rank for easier comparison
comparison = comparison.sort_values("Cosine Rank").reset_index(drop=True)

display(comparison.round({
    "Cosine Score": 4,
    "Dot Product Score": 4
}))

```
