```python
chunks = [
    "The Amazon rainforest contains millions of species of plants and animals.",
    "Climate change is increasing global temperatures and sea levels.",
    "Photosynthesis converts sunlight into chemical energy in plants.",
    "Renewable energy sources include solar, wind, and hydroelectric power.",
    "Electric vehicles help reduce greenhouse gas emissions."
]
len(chunks)
```

# Create Embeddding Vector for all chunks and query 

```python
import numpy as np
from sentence_transformers import SentenceTransformer

# other available models - all-MiniLM-L6-v2 with bge-large-en-v1.5, e5-large-v2, or text-embedding-3-large.
model = SentenceTransformer("all-MiniLM-L6-v2")

query = "How do plants produce energy?"


chunks_embeddings = model.encode(chunks)
query_embeddings = model.encode([query])
```

```python
print(chunks_embeddings.shape, query_embeddings.shape)
# print(chunks_embeddings[0])
```

# Cosine Similarity 

```python
from sklearn.metrics.pairwise import cosine_similarity 

scores = cosine_similarity(query_embeddings, chunks_embeddings)[0]
# print(scores)
rankings = np.argsort(scores)[::-1]

print("Question :", query, "\n")
for rank, indx in enumerate(rankings):
    print(f"Rank : {rank+1}")
    print(f"Score : {scores[indx]}")
    print(chunks[indx])
    print()
```

# Dot Product 

```python
import numpy as np 
scores = np.dot(chunks_embeddings, query_embeddings[0])
print(scores)
rankings = np.argsort(scores)[::-1]

print("Question :", query, "\n")
for rank, indx in enumerate(rankings):
    print(f"Rank : {rank+1}")
    print(f"Score : {scores[indx]}")
    print(chunks[indx])
    print()
```

# Euclidean Distance 

```python
from sklearn.metrics.pairwise import euclidean_distances

distances = euclidean_distances(query_embeddings, chunks_embeddings)[0]
rankings = np.argsort(distances)

print("Question :", query, "\n")
for rank, indx in enumerate(rankings):
    print(f"Rank : {rank+1}")
    print(f"Score : {distances[indx]}")
    print(chunks[indx])
    print()
```

# Manhatten Distance 

```python
from sklearn.metrics.pairwise import manhattan_distances

distances = manhattan_distances(query_embeddings, chunks_embeddings)[0]
rankings = np.argsort(distances)

print("Question :", query, "\n")
for rank, indx in enumerate(rankings):
    print(f"Rank : {rank+1}")
    print(f"Score : {distances[indx]}")
    print(chunks[indx])
    print()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.metrics.pairwise import manhattan_distances
from sklearn.metrics.pairwise import euclidean_distances
import numpy as np 

model = SentenceTransformer("all-MiniLM-L6-v2")
def get_top_match(query):
    results = {}
    query_embeddings = model.encode([query])

    # Compute cosine similarity 
    scores = cosine_similarity(query_embeddings, chunks_embeddings)[0]
    rankings = np.argsort(scores)[::-1]
    results["Cosine"] = chunks[rankings[0]]

    # Compute dot product 
    scores = np.dot(chunks_embeddings, query_embeddings[0])
    rankings = np.argsort(scores)[::-1]
    results["Dot Product"] = chunks[rankings[0]]

    # Manhtten Distance 
    distances = manhattan_distances(query_embeddings, chunks_embeddings)[0]
    rankings = np.argsort(distances)
    results["Man Hatten"] = chunks[rankings[0]]

    # Euclidean Distance 
    distances = euclidean_distances(query_embeddings, chunks_embeddings)[0]
    rankings = np.argsort(distances)
    results["Euclidean"] = chunks[rankings[0]]

    return results
```
