```python
from pydantic import BaseModel, Field

class Employee(BaseModel):
    uid : int = Field(gt=0, lt=100)
    name : str
    primaryAddress : str
    secondaryAddress : str = None
```

```python
emp = Employee(uid = 1, name = "vivek", primaryAddress = "Moodbidri")
```

```python
print(emp)
```

```python
emp.model_json_schema()
```

```python
emp.model_dump_json()
```

```python
emp.model_dump()
```
