# 1. create an object and access

```python
from pydantic import BaseModel

class User(BaseModel):
  uid : int
  name : str

sample = User(uid = 10 , name = "Suraj")
print(sample)
print(sample.uid, type(sample.uid))
print(sample.name, type(sample.name))
```

# 2. auto type casting

```python

sample = User(uid = "10" , name = "Suraj")
print(sample)
print(sample.uid, type(sample.uid))
print(sample.name, type(sample.name))
```

# 5. Optional Fields

```python

from pydantic import BaseModel
from typing import Optional

class User(BaseModel):
  uid : int
  name : str
  # email : Optional[str] = None
  email : str = None
```

```python
sample = User(uid =10, name="suraj", email="sample@gmail.com" ) # pass all values
print(sample)

sample = User(uid =10, name="suraj") # Do not send optional value
print(sample)
```

# 6. Data Validation

```python
from pydantic import BaseModel, Field
class User(BaseModel):
  age : int = Field(gt=17, lt=40)
  name : str = Field(min_length=3, max_length=20)
  # email : Optional[str] = None
  email : str = None

sample = User(age =10, name="suraj", email="sample@gmail.com")
print(sample)
```

```python
sample = User(age =20, name="suraj", email="sample@gmail.com")
print(sample)
```

```python
# string validation example
sample = User(age =20, name="AK", email="sample@gmail.com")
print(sample)
```

```python
sample = User(age =20, name="suraj", email="sample@gmail.com")
print(sample)
```

# 7. Custom Data Validation

```python
from pydantic import BaseModel, Field, field_validator
class User(BaseModel):
  age : int = Field(gt=17, lt=40)
  name : str = Field(min_length=3, max_length=20)
  # email : Optional[str] = None
  email : str = None

  @field_validator("email")
  @classmethod
  def email_data_validator(cls,value):
    if not value.endswith("@microdegree.com"):
      raise ValueError("Sorry Your not eligible for creating this object")
    return value
```

```python
# with wrong input
sample = User(age =20, name="suraj", email="sample@gmail.com")
print(sample)
```

```python
# with correct input
sample = User(age =20, name="suraj", email="sample@microdegree.com")
print(sample)
```

# 7. Supports Description for each field

```python
from pydantic import BaseModel, Field

class User(BaseModel):
  age : int = Field(gt=17, lt=40, description="Agen of an employee")
  name : str = Field(min_length=3, max_length=20 ,description= "Full name of an employee")
  # email : Optional[str] = None
  email : str = None
```

# 8. can convert to dictionary and json

```python
# json schema
User.model_json_schema()
```

```python
# Convert object into json string
sample = User(age =20, name="suraj", email="sample@microdegree.com")
print(sample, type(sample))

jdata = sample.model_dump_json()
print(jdata, type(jdata))
```

```python
# Convert object into dictionary
sample = User(age =20, name="suraj", email="sample@microdegree.com")
print(sample, type(sample))

jdata = sample.model_dump()
print(jdata, type(jdata))
```

```python
print(jdata["age"], jdata["name"], jdata["email"])
```
