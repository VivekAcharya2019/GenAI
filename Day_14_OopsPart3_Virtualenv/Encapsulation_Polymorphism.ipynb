# 2. Encapsulation 

    ## Attributes
     - public 
     - private 
     - protected

## Public Attribute 

```python
class SmartDoor:
    def __init__(self):
        self.status = "locked"

    def lock(self):
        self.status = "locked"

    def unlock(self):
        self.status = "unlocked"
```

```python
# Public Attribute
door1 = SmartDoor()
print(door1.status)  # read 
door1.status = "unlocked"  # Modify 
print(door1.status)  # read 
```

## Private Attribute/Variable

```python
class SmartDoor:
    def __init__(self):
        self.__status = "locked"

    def lock(self):
        self.__status = "locked"

    def unlock(self):
        self.__status = "unlocked"

    def show_status(self):
       return self.__status
```

```python
# Private Attribute
door1 = SmartDoor()
print(door1.__status)  # read 
# door1.__status = "unlocked"  # Modify 
# print(door1.__status)  # read 
```

```python
door1.show_status()
```

```python
door1 = SmartDoor()
print(door1.show_status())
door1.lock() # Modify 
print(door1.show_status())  # read 
door1.unlock() # Modify 
print(door1.show_status())  # read 
```

## Protected Attribute/Variable

    python will not restrict it but it is discouraged

```python
class SmartDoor:
    def __init__(self):
        self._status = "locked"

    def lock(self):
        self._status = "locked"

    def unlock(self):
        self._status = "unlocked"

    def show_status(self):
       return self._status
```

```python
# Protcted Attribute
door1 = SmartDoor()
print(door1._status)  # read 
door1._status = "outside unlocked "  # Modify 
print(door1._status)  # read 
```

# 3. Polymorphism 

```python
def add(a,b):
    return a+b 

def add(a,b,c):
    return a+b+c

print(add(10,20))
```

```python
class Payment:
    def process(self):
        pass

class UPI(Payment):
    def process(self):
        print("Payement processed with UPI")

class PhonePay(UPI):
    def process(self):
        print("Payement processed with UPI in PhonePay")
```

```python
upi_obj = UPI()
upi_obj.process()
```

```python
app_obj = PhonePay()
app_obj.process()
```
