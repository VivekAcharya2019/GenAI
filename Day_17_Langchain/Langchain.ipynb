```python
# json 
from langchain_core.output_parsers import JsonOutputParser
from langchain_ollama import OllamaLLM
from langchain_core.prompts import PromptTemplate

# Connect to model 
llm = OllamaLLM(model = "llama3.2:1b")

# create a template 
prompt = PromptTemplate.from_template("""
        
        You are a  comment reviwer , review below comment and create response in json format 

        comment : {comment}

        return json only 
        format : 
        {{"classified_as" : "Positive/Negative/Neutral"
        "reply_message" " ""," }}

        
    """)

# create json parser object 
parser = JsonOutputParser()


#chain = prompt.invoke() -> llm.invoke()
chain = prompt | llm


```

```python
comment =  "Your product was bad "
response = chain.invoke({"comment":comment})
print(response, type(response))
response_after_parseing = parser.invoke(response)
print(response_after_parseing)
```

```python
type(response_after_parseing), response_after_parseing["classified_as"], response_after_parseing["reply_message"]
```

```python
comment =  "Product is verry good! will buy again"
response = chain.invoke({"comment":comment})
print(type(response),response)
response_after_parseing = parser.invoke(response)
print(type(response_after_parseing),response_after_parseing)
```

```python
# json 
from langchain_core.output_parsers import JsonOutputParser
from langchain_ollama import OllamaLLM
from langchain_core.prompts import PromptTemplate

# Connect to model 
llm = OllamaLLM(model = "llama3.2:1b")

# create a template 
prompt = PromptTemplate.from_template("""
        
        You are a  comment reviwer , review below comment and create response in json format 

        comment : {comment}

        return json only 
        format : 
        {{"classified_as" : "Positive/Negative/Neutral"
        "reply_message" " ""," }}

        
    """)

# create json parser object 
parser = JsonOutputParser()


#chain = prompt.invoke() -> llm.invoke()
chain = prompt | llm | parser
```

```python
comment =  "Product is verry good! will buy again"
response = chain.invoke({"comment":comment})
print(type(response), response)
```
