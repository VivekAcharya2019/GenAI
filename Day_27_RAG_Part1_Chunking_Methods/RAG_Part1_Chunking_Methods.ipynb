```python
def show_chunks(chunks):
    print("TOTAL CHUNKS ", len(chunks))
    for i , chunk in enumerate(chunks):
        print("=="*40)
        print(f"CHUNK No. {i} : Chunk Size : {len(chunk)}")
        print("_"*20)
        print(chunk)
```

# Fixed Size Chunking
    - Chunks based on fixed size 
    - scarped data 

    cons 
    - No connection between chunks 
    - No gramatical borders are considered , words can break 


```python
def fixed_size_chunking(document, chunk_size):
    chunks = []
    for i in range(0, len(document),chunk_size):
        chunk = document[i : i+chunk_size]
        chunks.append(chunk)

    return chunks

chunk_size = 200
chunks = fixed_size_chunking(document, chunk_size)
show_chunks(chunks)
```

# Fixed Chunking with Overlapping 

```python
def fixed_chunking_with_overlap(document, chunk_size, overlap):
    chunks = []
    start = 0 
    while start< len(document):
        chunk = document[start: start+chunk_size]
        chunks.append(chunk)
        start = start + chunk_size - overlap
    return chunks

chunk_size = 250 
overlap = 50
chunks = fixed_chunking_with_overlap(document, chunk_size, overlap)
show_chunks(chunks)
```

# Sentence Based chunking 
    - creates chunks based on sentence 

```python
import nltk 
from nltk.tokenize import sent_tokenize

chunks = sent_tokenize(document)
show_chunks(chunks)
```

```python
def overlap_sent_chunking(document, sentence_count_per_chunk, overlapping):
    sent_chunks = sent_tokenize(document)
    start = 0 
    chunks = []
    while start< len(sent_chunks):
        chunk = "\n".join(sent_chunks[start:start+sentence_count_per_chunk])
        start = start + sentence_count_per_chunk - overlapping
        chunks.append(chunk)
    return chunks
    
sentence_count_per_chunk = 3
overlapping = 1
chunks = overlap_sent_chunking(document, sentence_count_per_chunk, overlapping)
show_chunks(chunks)
```

# 1. Recursive Character Text Splitting 
 - splits text recursively using a list of seperators untill chunk size is reached 

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size = 400,
                                           chunk_overlap = 100 , 
                                           separators = ["\n\n", "\n", "."]
                                          )

chunks = text_splitter.split_text(document)
show_chunks(chunks)
```

```python
code = """
# load model for chat bot 
from langchain_ollama import ChatOllama
from langchain_core.tools import tool 
from langchain_core.prompts import MessagesPlaceholder, ChatPromptTemplate


## Create all tools 
@tool
def square(x :int) -> int:
    return x**2

@tool
def add(a :int, b:int) -> int:
    return a+b

@tool
def weather(city:str)->str:
    # weather = requests.get("")
    return "rainy weather is predicted"


# Initialize model and bind all tools 
tools = [square, add]
chat_model = ChatOllama(model="llama3.2:1b")
chat_model_with_tools = chat_model.bind_tools(tools)


# Create Prompt template with history placeholder
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpfull assistant."),
    MessagesPlaceholder(variable_name="history"),   ## All history message 
    ("human", "{input}")
]
)

# Create a chain to connect prompt  and model 
chat_chain = prompt |  chat_model
history = []  # Create a empty list to save the conversation history

# In Loop Chat with model 
while True:
    # Get user message 
    user_message = input("\n User : ")

    # break loop if user says q
    if user_message == "q":
        break 

    # .invoke the chain with user message and history 
    response = chat_chain.invoke({"input" : user_message,"history" : history})

    # Print the AI message 
    print("\n Bot : ", response.content)

    # append both user and ai message to history
    history.append(("human", user_message))
    history.append(("ai", response.content))

"""
```
