## Methods

```python
class User:
    # Constuctor 
    def __init__(self, name , amount, branch="Bangalore"):
        self.name = name
        self.balance = amount
        self.branch = branch

    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self, amount):
        self.balance = self.balance - amount 

    def show_details(self):
        print("Existing Balance", self.balance)
```

```python
user1 = User("Viresh", 10000 ,"Mangalore" )  # 
user2 = User("Vinay", 20000 )  # 
```

```python
## Accessing a methods
# By default first argument will be object itself 
user1.show_details()
user1.deposit(9999)
user1.show_details()
user1.withdraw(9999)
user1.show_details()
```

## Dunder methods   ___
    1. __init__ -- When creating object
    2. __str__  -- when print like functions are called

```python
class User:
    # Constuctor 
    def __init__(self, name , amount, branch="Bangalore"):
        self.name = name
        self.balance = amount
        self.branch = branch
        self.loan_amount = 1200

    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self, amount):
        self.balance = self.balance - amount 

    def show_details(self):
        print("Existing Balance", self.balance)

    def __str__(self):
        print("__str__ method is called")
        return f" {self.name=}  {self.branch=}"

    def __len__(self):
        print("__len__ method is called")
        return self.balance

    def __gt__(self, other):
        print("__gt__ method is called")
        return self.balance > other.balance

    def __eq__(self, other):
        return (self.name == other.name) and (self.balance == other.balance)
```

```python
user1 = User("Viresh", 10000 ,"Mangalore" )  # 
user2 = User("Vinay", 20000 )  # 
user3 = User("Vinay", 20000 )  # 
print(user1)
```
