# 1. Inheritance

```python
class Car:
    def __init__(self, name, engine, cc):
        self.name = name
        self.engine = engine
        self.cc = cc 

    def show_car(self):
        return f"{self.name=} {self.cc=}"
```

```python
v1car = Car("TATA",6, 60000)
v1car.show_car()
```

# 1. Single Inheritance

```python
#     ChildClass(ParentClass)
# ChildClass constructor will be called by default
# manually have to call the parent class constructor 
class Car2(Car):
    def __init__(self, name, engine, cc,adas):
        # Car.__init__(self,name, engine, cc)
        super().__init__(name, engine, cc)
        self.adas = adas

    def show_car(self):
        print("This is inside the child class")
        return f"{self.name=} {self.cc=}"
```

```python
v2car = Car2("TATA",6, 60000,"8camers")
v2car.show_car()
```

```python
v2car.show_car()
```

# 2. Multiple Inheritance 

```python
class Department:
    def __init__(self, eid, department):
        self.eid = eid
        self.department = department

    def show_department(self):
        print("Depart Is ", self.department)

    def get_id(self):
        return self.eid

    def __str__(self):
        return "This is department class object"

class Developer:
    def __init__(self, language,experience):
        self.language = language 
        self.experience = experience

    def get_langauge(self):
        return self.language

class Employee(Department,Developer):
    def __init__(self, name, eid, department,language , experience ):
        Department.__init__(self,eid, department)
        Developer.__init__(self,language , experience)
        self.name = name

    def show_experience(self):
        print(self.experience)
        
```

```python
emp = Employee("Vikas", "E001", "MLOPS", "Python", 10)
```

```python
emp.show_experience()
```

```python
emp.show_department()
```

```python
emp.get_langauge()
```

```python
print(emp)
```

```python
Employee.__mro__
```

# 3. MultiLevel Inheritance

```python
class Department:
    def __init__(self, eid, department):
        self.eid = eid
        self.department = department

    def show_department(self):
        print("Depart Is ", self.department)

    def get_id(self):
        return self.eid

    def __str__(self):
        return "This is department class object"

class Developer(Department):
    def __init__(self, eid, department,language,experience, ):
        super().__init__(eid, department)
        self.language = language 
        self.experience = experience

    def get_langauge(self):
        return self.language

class Employee(Developer):
    def __init__(self, name, eid, department,language , experience ):
        super().__init__(eid, department,language , experience)
        self.name = name

    def show_experience(self):
        print(self.experience)

    def get_langauge(self):
        pass
```
