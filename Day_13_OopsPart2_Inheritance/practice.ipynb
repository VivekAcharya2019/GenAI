```python
#create sample example for multiple inheritence in python
class A:
    def __init__(self):
        self.name = "Class A"

    def method_a(self):
        print(f"Method A from class A: {self.name}")

class B:
    def __init__(self):
        self.name = "Class B"

    def method_b(self):
        print(f"Method B from class B: {self.name}")

class C(B, A):
    def __init__(self):
        self.name = "Class C"
        A.__init__(self)
        B.__init__(self)

    def method_c(self):
        print(f"Method C from class C: {self.name}")

# Create an instance of class C
c_instance = C()
# Call methods from class A, B, and C
c_instance.method_a()  # Output: Method A from class A  
c_instance.method_b()  # Output: Method B from class B
c_instance.method_c()  # Output: Method C from class C
print(c_instance.name)  # Output: Class C

print(c_instance.__class__.__mro__)  # Output: (<class '__main__.C'>, <class '__main__.A'>, <class '__main__.B'>, <class 'object'>)
```
