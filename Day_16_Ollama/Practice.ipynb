# API Endpoint for chat like api calls 

```python
import requests

url = "http://127.0.0.1:11434/api/chat"

history = [
{
    "role" : "system",
    "content" : "You are a helpful assistant"
},
  {
      "role": "user",
      "content": "what is 2+2?"
  }
]

payload = {
    "model": "qwen3.5:0.8b",
    "messages": history,
    "stream" : False
}

response = requests.post(url, json = payload)
jdata = response.json()
print(jdata)
```

```python
import requests

url = "http://127.0.0.1:11434/api/chat"
history = [
    {
    "role" : "system",
    "content":"You are helfull assistant"
    }
]

while True:

    user_query = input("User:")

    if(user_query == "q"):
        break

    user_msg = {
        "role":"user",
        "content" : user_query
    }

    history.append(user_msg)

    payload = {
        "model" : "qwen3.5:0.8b",
        "messages": history,
        "stream" : False
    }
    for item in history:
        print (f"{item['role']} : {item['content']}")
    response = requests.post(url, json=payload)
    jdata = response.json()

    history.append(jdata["message"])
    
    print("Bot:", jdata["message"]["content"])
```

```python
from ollama import chat

history = []

while True:
    user_query = input("User:")

    if(user_query == "q"):
        break
        
    user_msg = {
    "role" : "user",
    "content" : user_query
    }

    history.append(user_msg)

    jdata = chat(model="qwen3.5:0.8b",messages=history)

    history.append(jdata["message"])

    print("\n Bot : ",jdata["message"]["content"])
                   
    
```
