```python
from langchain_ollama import OllamaEmbeddings
em_model = OllamaEmbeddings(model="embeddinggemma-2")

chunk_embeddings = em_model.embed_documents(sample_chunks)
```

# Convert chunk embeddings into float 32 data types using numpy

```python
import numpy as np

chunk_embeddings = np.asarray(chunk_embeddings, dtype=np.float32)
```

# create FAISS index and add chuncks to that

```python
import faiss

dimension = chunk_embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
index.add(chunk_embeddings)
```

# generate 1 valid question to query the chunks and embedd it, query the index get 3 top matching chunks


```python
# Generate one valid question and embed it for semantic search
question = "What are the main benefits of renewable energy for the environment and economy?"
query_embedding = em_model.embed_query(question)
query_embedding = np.asarray(query_embedding, dtype=np.float32).reshape(1, -1)

# Search the FAISS index for the most relevant chunks
distances, indices = index.search(query_embedding, 3)

# Print the top matching chunks
for i, idx in enumerate(indices[0]):
    print(f"Match {i+1} (distance={distances[0][i]}):")
    print(sample_chunks[idx])
    print("-" * 80)
```
