```python
# Initialize the llm model using langchain
from langchain_ollama import OllamaLLM

llm = OllamaLLM(model="llama3.2:1b")
```

## 6. Chain of Thoughts

```python
prompt = """
A train is travelling at 60km/h
It will stop for 10 mins in 1 hour
how far it will travell in 3.5 hours without stopping on any station?

"""

response = llm.invoke(prompt)
print(response)
```

```python
prompt = """
A train is travelling at 60km/h
It will stop for 10 mins in 1 hour
how far it will travell in 3.5 hours without stopping on any station?

think step by step
"""

response = llm.invoke(prompt)
print(response)
```

## 7. Self- Critique Prompting

```python
prompt = """
A train is travelling at 60km/h
It will stop for 10 mins in 1 hour
how far it will travell in 3.5 hours without stopping on any station?

think step by step
then review your answer for mistakes
Provide correct answer
"""

response = llm.invoke(prompt)
print(response)
```
