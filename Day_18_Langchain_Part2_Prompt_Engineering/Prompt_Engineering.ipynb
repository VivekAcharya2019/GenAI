# Load Model for testing different prompts

```python
from langchain_ollama import OllamaLLM

# Connect to model 
llm = OllamaLLM(model = "llama3.2:1b")
```

# 1. Basic Prompting  

```python
response = llm.invoke("Explain Python")
print(response)
```

# 2. Role Prompting 

```python
prompt  = """
    You are a mechanical engineer 
    answer below user question 

    user_q : "Explain Transformer"
    
"""
response = llm.invoke(prompt)
print(response)
```

```python
prompt  = """
    You are a Computer Science student learning LLM  
    your role is to answer any question in 10 words
    answer below user question
    user_q : "Explain Transformer"
    
"""
response = llm.invoke(prompt)
print(response)
```

## 3. Instruction Prompts 

```python
prompt  = """
        Explain Transformer 

        Requirements : 
           - Beginner friendly 
           - Less than 40 words 
           - include analogy
    
"""
response = llm.invoke(prompt)
print(response)
```

## 4. Delimited prompts 

```python
prompt  = """
        Explain Transformer 
        
        -----------------------------------------------
        Requirements : 
           - Beginner friendly - Less than 40 words 
           - include analogy
        -----------------------------------------------
    
"""
response = llm.invoke(prompt)
print(response)
```

## 5 Few Shot Prompting 

    Classification 
    Formatting 
    Extraction 
    Transformation 
    

```python
comment =  "produc ran perfectly for only 5 days"
prompt  = f"""
        You are a helpfull assistant reviewing user comments provide sentiment and response message.

        Example : 
            user_message = "your product quality is bad"
            response : {{
                        review : "Negative",
                        message : "Sorry to hear that can you please let us know the purchase date"
                        }}

            user_message = "your product quality is too good"
            response : {{
                        review : "Positive",
                        message : "Thanks please visit again"
                        }}

        comment  : {comment}
    
"""
response = llm.invoke(prompt)
print(response)
```
