```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

system_msg = SystemMessage(content="You are helpfull assistant")
messages = [ system_msg ]

while True:
    user_message = input("User : ")
    if user_message == "q":
        aimessage = AIMessage("Thanks for the chat! Please come again")
        print("Bot : ",aimessage.content)
        break
        

    # wrap the string as HumanMessage
    hmessage = HumanMessage(content=user_message)
    messages.append(hmessage)

    # Send all messages and get response 
    ai_message = chat_model.invoke(messages)
    messages.append(ai_message)
    print("Bot : ",ai_message.content)

    
```
